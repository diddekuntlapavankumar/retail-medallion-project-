```sql
%sql
CREATE CATALOG IF NOT EXISTS retail_lakehouse;

CREATE SCHEMA IF NOT EXISTS retail_lakehouse.config;
CREATE SCHEMA IF NOT EXISTS retail_lakehouse.bronze;
CREATE SCHEMA IF NOT EXISTS retail_lakehouse.silver;
CREATE SCHEMA IF NOT EXISTS retail_lakehouse.gold;
CREATE SCHEMA IF NOT EXISTS retail_lakehouse.audit;
```

```sql
show catalogs;
```

```sql
CREATE TABLE IF NOT EXISTS retail_lakehouse.config.pipeline_config (
  pipeline_id STRING,
  source_name STRING,
  source_path STRING,
  target_table STRING,
  load_type STRING,
  watermark_column STRING,
  active_flag BOOLEAN
);

CREATE TABLE IF NOT EXISTS retail_lakehouse.audit.watermark_control (
  pipeline_id STRING,
  last_watermark_value TIMESTAMP
);

CREATE TABLE IF NOT EXISTS retail_lakehouse.audit.pipeline_audit (
  run_id STRING,
  pipeline_id STRING,
  layer_name STRING,
  status STRING,
  start_time TIMESTAMP,
  end_time TIMESTAMP,
  records_read BIGINT,
  records_written BIGINT,
  error_message STRING
);
```
